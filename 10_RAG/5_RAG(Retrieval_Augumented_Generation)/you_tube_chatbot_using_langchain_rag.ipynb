```python
from youtube_transcript_api import YouTubeTranscriptApi
from youtube_transcript_api._errors import TranscriptsDisabled
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings, ChatGoogleGenerativeAI
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
```

# Step 1(A): Indexing (Document Ingestion)


```python
video_id = "c64hqovEG-U" # only the ID, not full URL

try:
    # Create API object
    ytt_api = YouTubeTranscriptApi()

    # Fetch transcript (English)
    transcript_list = ytt_api.fetch(video_id, languages=["en"])

    # Convert to plain text
    transcript = " ".join(chunk.text for chunk in transcript_list)

    print(transcript)

except TranscriptsDisabled:
    print("No captions available for this video.")
```

```python
for i in transcript_list:
    print(i)
```

# Step 1(B): Indexing (Text Splitting)


```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
chunks = splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
chunks[7]
```

# Step 1(C) and 1(D): Indexing (Embedding Generation and Storing in Vector Store)


```python
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")
vector_store = FAISS.from_documents(chunks, embeddings)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(['e20410d1-6155-4fba-a46d-78b9559f5cf0'])
```

# Step 2: Retrieval


```python
retriever = vector_store.as_retriever(search_type="similarity", search_kwargs={"k": 5})
```

```python
retriever
```

```python
retriever.invoke('Tell me about Spinning Stars & Gyrochronology')
```

# Step 3: Augumentation


```python
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0.2)
```

```python
prompt = PromptTemplate(
    template="""
      You are a helpful assistant.
      Answer ONLY from the provided transcript context.
      If the context is insufficient, just say you don't know.

      {context}
      Question: {question}
    """,
    input_variables = ['context', 'question']
)
```

```python
question = "Explain about Spinning Stars & Gyrochronology that discussed in this video ?"
retrieved_docs = retriever.invoke(question)
```

```python
retrieved_docs
```

```python
context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
context_text
```

```python
final_prompt = prompt.invoke({"context": context_text, "question": question})
```

```python
final_prompt
```

# Step 4: Generation


```python
answer = llm.invoke(final_prompt)
print(answer.content)
```

# Building Chain


```python
from langchain_core.runnables import RunnableParallel, RunnablePassthrough, RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
def format_docs(retrieved_docs):
  context_text = "\n\n".join(doc.page_content for doc in retrieved_docs)
  return context_text
```

```python
parallel_chain = RunnableParallel({
    'context': retriever | RunnableLambda(format_docs),
    'question': RunnablePassthrough()
})
```

```python
parallel_chain.invoke("Where Are The Sun's Siblings?")
```

```python
parser = StrOutputParser()
```

```python
main_chain = parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke('Can you summarize the video')
```
